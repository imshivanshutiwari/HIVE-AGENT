## 1. Configuration

## 2. Fetch Repository

```python
from github.repo_fetcher import RepoFetcher

# Replace with a public repository
REPO = 'python/cpython'
fetcher = RepoFetcher(token=os.getenv('GITHUB_TOKEN', ''))
repo_data = fetcher.fetch(REPO, max_files=20)
print(f"Fetched {len(repo_data.get('files', []))} files from {REPO}")
```

## 3. AST Analysis

```python
from analysis.ast_parser import ASTParser

parser = ASTParser()
sample_code = '''
def greet(name: str) -> str:
    """Return a greeting string."""
    return f"Hello, {name}!"

class Greeter:
    def __init__(self, prefix: str = "Hello"):
        self.prefix = prefix

    def greet(self, name: str) -> str:
        return f"{self.prefix}, {name}!"
'''

result = parser.parse(sample_code, language='python')
print('Functions:', [f['name'] for f in result.get('functions', [])])
print('Classes:',   [c['name'] for c in result.get('classes', [])])
```

## 4. Complexity Scoring

```python
from analysis.complexity_scorer import ComplexityScorer

scorer = ComplexityScorer()
scores = scorer.score(sample_code)
print('Complexity scores:', scores)
```
